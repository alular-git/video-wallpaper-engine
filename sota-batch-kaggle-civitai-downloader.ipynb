```python
from IPython.display import Audio, display
import numpy as np
import time
import threading
import warnings

warnings.filterwarnings("ignore", category=RuntimeWarning)

def play_silence():
    while True:
        silence = np.zeros(44100)
        display(Audio(silence, rate=44100, autoplay=True))
        time.sleep(60)

thread = threading.Thread(target=play_silence, daemon=True)
thread.start()
print("🎵 防断连音频已启动！请确保浏览器标签页未静音。")
```

```python
import urllib.parse
import re

# ================= ️ 在此处直接 Ctrl+V 粘贴您的批量 URL ⬇️ =================
# 规则：每行一个完整的 URL。可以留空行，可以用 # 开头写注释。
BATCH_URLS = """
# 示例：第一个模型
https://b2.civitai.com/file/civitai-modelfiles/model/8844536/sfbehindV21LowNoise.9hLl.safetensors?Authorization=3_20260710040512_7372ecc6f58b8d0cc7408a7a_34d9670d7e0e590b9cf622617e7b14b2ae9e5eb0_004_20260710050512_0050_dnld&b2ContentDisposition=attachment%3B+filename%3D%22sfbehind_v2.1_low_noise.safetensors%22
https://b2.civitai.com/file/civitai-modelfiles/model/8844536/sfbehindV21HighNoise.nJPT.safetensors?Authorization=3_20260710032639_14bc21bd00d37db167297d5f_f01eff64216722cdb516b47dc15d1c65e5ba930b_004_20260710042639_0051_dnld&b2ContentDisposition=attachment%3B+filename%3D%22sfbehind_v2.1_high_noise.safetensors%22

"""
# ========================================================================

#  自动解析逻辑
urls = []
filenames = []

for line in BATCH_URLS.strip().split('\n'):
    line = line.strip()
    # 跳过空行和注释
    if not line or line.startswith('#'):
        continue
    
    urls.append(line)
    
    # 自动解析文件名
    parsed = urllib.parse.urlparse(line)
    query = urllib.parse.parse_qs(parsed.query)
    
    if 'b2ContentDisposition' in query:
        disp = urllib.parse.unquote(query['b2ContentDisposition'][0])
        match = re.search(r'filename="([^"]+)"', disp)
        fname = match.group(1) if match else "model_unknown.safetensors"
    else:
        fname = parsed.path.split('/')[-1]
        
    filenames.append(fname)

print(f"✅ 成功解析 {len(urls)} 个下载任务：")
for i, (url, fname) in enumerate(zip(urls, filenames), 1):
    print(f"  [{i}] {fname}")
print(f"\n💡 提示：运行下一个单元格将开始批量下载（保留完整进度显示）")
```

```python
import os
import subprocess
import sys
import time

REMOTE_PATH = "Onedrive:AI-Models/WAN2.2/"
WORK_DIR = "/kaggle/working"
MIN_SPACE_GB = 15  # 安全阈值

def check_space():
    """检查磁盘空间"""
    stat = os.statvfs(WORK_DIR)
    free_gb = (stat.f_bavail * stat.f_frsize) / 1073741824
    print(f"💾 当前磁盘可用空间: {free_gb:.1f} GB")
    if free_gb < MIN_SPACE_GB:
        raise Exception(f"️ 磁盘空间不足 ({free_gb:.1f}GB < {MIN_SPACE_GB}GB)！")

def download_with_aria2(url, output):
    """使用 aria2c 下载（保留原有进度显示）"""
    print(f"\n⬇️ 开始下载: {output}")
    print("⚡ 使用 aria2c 8 线程下载...")
    
    cmd = [
        "aria2c", "-x", "8", "-s", "8", "-k", "1M", "-c",
        "--header=User-Agent: Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36",
        "--header=Referer: https://civitai.com/",
        "-d", WORK_DIR,
        "-o", output,
        url
    ]
    
    result = subprocess.run(cmd, capture_output=False, text=True)
    return os.path.exists(os.path.join(WORK_DIR, output))

def upload_with_rclone(filepath):
    """使用 rclone 上传（保留原有文本进度条）"""
    full_path = os.path.join(WORK_DIR, filepath)
    print(f"\n🚀 开始上传: {filepath} 到 {REMOTE_PATH}")
    print("⏳ 开始上传，实时进度如下...")
    print("="*60)
    
    cmd = [
        "rclone", "move", full_path, REMOTE_PATH,
        "--progress",
        "--transfers", "4",
        "--buffer-size", "128M",
        "--stats", "5s",
        "--retries", "3"
    ]
    
    start_time = time.time()
    
    # 实时读取 rclone 的输出（保留原有逐行打印方式）
    process = subprocess.Popen(
        cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )
    
    for line in process.stdout:
        line = line.strip()
        if line:
            print(line)
            sys.stdout.flush()
    
    process.wait()
    elapsed = time.time() - start_time
    
    print("="*60)
    
    success = process.returncode == 0 and not os.path.exists(full_path)
    
    if success:
        print(f"✅ 上传成功！耗时: {elapsed/60:.1f} 分钟")
    elif process.returncode == 0:
        print(f"⚠️ 上传完成，但本地文件未被删除（可能 OneDrive 已存在同名文件）")
    else:
        print(f"❌ 上传中断或失败，退出码: {process.returncode}")
    
    return success

# 🚀 主循环
print(f"🚀 启动批量流水线，共 {len(urls)} 个任务")
print(f"📁 目标路径: {REMOTE_PATH}")
print("="*60)

success_count = 0
start_total = time.time()

for i, (url, fname) in enumerate(zip(urls, filenames), 1):
    print(f"\n{'#'*60}")
    print(f"📦 任务进度: [{i}/{len(urls)}]")
    print(f"🎯 当前文件: {fname}")
    print(f"{'#'*60}")
    
    try:
        # 1. 检查空间
        check_space()
        
        # 2. 下载（显示 aria2c 进度）
        if not download_with_aria2(url, fname):
            print(f" [{i}] 下载失败，跳过此文件")
            continue
        
        # 3. 上传（显示 rclone 进度）
        if upload_with_rclone(fname):
            success_count += 1
            print(f"✅ [{i}/{len(urls)}] 完成！剩余任务: {len(urls) - i}")
        else:
            print(f"⚠️ [{i}] 上传可能有问题，请检查")
            
    except Exception as e:
        print(f"❌ 任务中断: {e}")
        print("💡 您可以重新运行此单元格继续未完成的任务")
        break

total_elapsed = time.time() - start_total
print(f"\n{'='*60}")
print(f"🎉 批量任务结束！")
print(f"✅ 成功完成: {success_count}/{len(urls)}")
print(f"⏱️  总耗时: {total_elapsed/60:.1f} 分钟")
print("💡 提示：回到单元格 3 粘贴新 URL 可继续下载")
```
